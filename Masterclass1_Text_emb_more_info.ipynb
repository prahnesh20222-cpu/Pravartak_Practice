# **CountVectorizer**

```python
from sklearn.feature_extraction.text import CountVectorizer
import numpy as np

sentences = [

    "Delhi is the capital of India.",
    "Tokyo is a major city in Japan.",
    "Paris is known as the city of lights.",
    "New York is one of the busiest cities in the world.",
    "London is famous for its historical landmarks.",
    "Sydney is a coastal city known for the Opera House.",
    "LLMs are used for text processing.",
    "Transformers help machines understand language.",
    "NLP models learn linguistic patterns.",
    "Tokenization converts text into numerical units.",
    "Language models generate human-like responses.",
    "Text classification assigns labels to sentences.",
    "CNNs are popular for image classification.",
    "Vision Transformers perform well on image tasks.",
    "Object detection identifies items in pictures.",
    "Image segmentation divides an image into meaningful regions.",
    "Face recognition systems detect and identify faces.",
    "Optical character recognition extracts text from images."
]


# ============================================================
# STEP 1 — CREATE COUNT VECTORIZER
# ============================================================

cv = CountVectorizer()


# ============================================================
# STEP 2 — LEARN THE VOCABULARY
# ============================================================

cv.fit(sentences)

print("=" * 80)
print("STEP 1: VOCABULARY")
print("=" * 80)

vocabulary = cv.vocabulary_

print("\nWord → Column Index\n")

for word, index in sorted(vocabulary.items(), key=lambda x: x[1]):
    print(f"{word:20s} → Column {index}")


# ============================================================
# STEP 3 — SHOW FEATURE NAMES
# ============================================================

feature_names = cv.get_feature_names_out()

print("\n" + "=" * 80)
print("STEP 2: COLUMN NAMES")
print("=" * 80)

print("\nEach column represents one vocabulary word:\n")

for i, word in enumerate(feature_names):
    print(f"Column {i:3d} → {word}")


# ============================================================
# STEP 4 — TRANSFORM SENTENCES
# ============================================================

emb_cv = cv.transform(sentences).toarray()


# ============================================================
# STEP 5 — SHOW EACH SENTENCE AND ITS WORDS
# ============================================================

print("\n" + "=" * 80)
print("STEP 3: SENTENCE → TOKENS")
print("=" * 80)

for i, sentence in enumerate(sentences):

    # CountVectorizer's analyzer shows the actual tokens
    tokens = cv.build_analyzer()(sentence)

    print(f"\nSentence {i}")
    print("Original :", sentence)
    print("Tokens   :", tokens)


# ============================================================
# STEP 6 — SHOW HOW FIRST SENTENCE BECOMES A VECTOR
# ============================================================

print("\n" + "=" * 80)
print("STEP 4: HOW ONE SENTENCE BECOMES A VECTOR")
print("=" * 80)

sentence_id = 0

print("\nSentence:")
print(sentences[sentence_id])

print("\nNon-zero words:")

for column, value in enumerate(emb_cv[sentence_id]):

    if value > 0:
        print(
            f"Word = {feature_names[column]:15s} "
            f"| Column = {column:3d} "
            f"| Count = {value}"
        )


# ============================================================
# STEP 7 — SHOW COMPLETE MATRIX
# ============================================================

print("\n" + "=" * 80)
print("STEP 5: COMPLETE COUNT VECTORIZER MATRIX")
print("=" * 80)

print("\nShape:", emb_cv.shape)

print("\nRows    =", emb_cv.shape[0], "sentences")
print("Columns =", emb_cv.shape[1], "vocabulary words")

print("\nMatrix:\n")
print(emb_cv)


# ============================================================
# STEP 8 — SHOW MATRIX WITH WORD NAMES
# ============================================================

print("\n" + "=" * 80)
print("STEP 6: COLUMN INDEX → WORD")
print("=" * 80)

print("\n")

for i, word in enumerate(feature_names):
    print(f"{i:3d}: {word}")


# ============================================================
# STEP 9 — SHOW ONLY NON-ZERO VALUES FOR EVERY SENTENCE
# ============================================================

print("\n" + "=" * 80)
print("STEP 7: WHAT EACH SENTENCE ACTUALLY CONTAINS")
print("=" * 80)

for i, sentence in enumerate(sentences):

    print(f"\nSentence {i}: {sentence}")

    for column, value in enumerate(emb_cv[i]):

        if value > 0:
            print(
                f"    {feature_names[column]:20s} "
                f"→ column {column:3d} "
                f"→ count {value}"
            )


# ============================================================
# STEP 10 — SPECIFICALLY CHECK DELHI
# ============================================================

print("\n" + "=" * 80)
print("STEP 8: WHAT HAPPENS TO 'DELHI'?")
print("=" * 80)

delhi_column = cv.vocabulary_.get("delhi")

print("\nDelhi's column index:", delhi_column)

print("\nDelhi appears in:")

for i, sentence in enumerate(sentences):

    count = emb_cv[i][delhi_column]

    if count > 0:
        print(
            f"Sentence {i}: "
            f"'{sentence}' "
            f"→ Delhi count = {count}"
        )


# ============================================================
# STEP 11 — SHOW THE FIRST SENTENCE AS WORD → COUNT
# ============================================================

print("\n" + "=" * 80)
print("STEP 9: FIRST SENTENCE → WORD COUNTS")
print("=" * 80)

first_sentence_vector = emb_cv[0]

for column, word in enumerate(feature_names):

    if first_sentence_vector[column] > 0:
        print(
            f"{word:15s} = "
            f"{first_sentence_vector[column]}"
        )
```

# **TfidVectorizer**

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import numpy as np

# ============================================================
# SENTENCES
# ============================================================

sentences = [

    "Delhi is the capital of India.",
    "Tokyo is a major city in Japan.",
    "Paris is known as the city of lights.",
    "New York is one of the busiest cities in the world.",
    "London is famous for its historical landmarks.",
    "Sydney is a coastal city known for the Opera House.",
    "LLMs are used for text processing.",
    "Transformers help machines understand language.",
    "NLP models learn linguistic patterns.",
    "Tokenization converts text into numerical units.",
    "Language models generate human-like responses.",
    "Text classification assigns labels to sentences.",
    "CNNs are popular for image classification.",
    "Vision Transformers perform well on image tasks.",
    "Object detection identifies items in pictures.",
    "Image segmentation divides an image into meaningful regions.",
    "Face recognition systems detect and identify faces.",
    "Optical character recognition extracts text from images."
]


# ============================================================
# STEP 1 — INITIALIZE TF-IDF VECTORIZER
# ============================================================

tfidf = TfidfVectorizer()

print("=" * 80)
print("STEP 1: INITIALIZE TF-IDF")
print("=" * 80)

print("""
TF-IDF converts text into numerical vectors.

TF  = Term Frequency
IDF = Inverse Document Frequency

Final value = TF × IDF
""")


# ============================================================
# STEP 2 — LEARN VOCABULARY
# ============================================================

tfidf.fit(sentences)

feature_names = tfidf.get_feature_names_out()
vocabulary = tfidf.vocabulary_

print("=" * 80)
print("STEP 2: VOCABULARY")
print("=" * 80)

print("\nWord → Column Index\n")

for word, index in sorted(vocabulary.items(), key=lambda x: x[1]):
    print(f"{word:20s} → Column {index}")


# ============================================================
# STEP 3 — SHOW TOKENS
# ============================================================

print("\n" + "=" * 80)
print("STEP 3: TOKENIZATION")
print("=" * 80)

analyzer = tfidf.build_analyzer()

for i, sentence in enumerate(sentences):

    tokens = analyzer(sentence)

    print(f"\nSentence {i}")
    print("Original :", sentence)
    print("Tokens   :", tokens)


# ============================================================
# STEP 4 — TRANSFORM SENTENCES
# ============================================================

emb_tfidf = tfidf.transform(sentences).toarray()

print("\n" + "=" * 80)
print("STEP 4: TF-IDF MATRIX")
print("=" * 80)

print("\nMatrix shape:", emb_tfidf.shape)

print("\nRows    =", emb_tfidf.shape[0], "sentences")
print("Columns =", emb_tfidf.shape[1], "vocabulary words")

print("\nTF-IDF Matrix:\n")

print(np.round(emb_tfidf, 4))


# ============================================================
# STEP 5 — SHOW COLUMN → WORD
# ============================================================

print("\n" + "=" * 80)
print("STEP 5: COLUMN → WORD")
print("=" * 80)

for i, word in enumerate(feature_names):

    print(
        f"Column {i:3d} → {word}"
    )


# ============================================================
# STEP 6 — SHOW NON-ZERO TF-IDF VALUES
# ============================================================

print("\n" + "=" * 80)
print("STEP 6: WORD → TF-IDF VALUE")
print("=" * 80)

for i, sentence in enumerate(sentences):

    print(f"\nSentence {i}:")
    print(sentence)

    print("\nNon-zero TF-IDF values:")

    for column, value in enumerate(emb_tfidf[i]):

        if value > 0:

            print(
                f"    {feature_names[column]:20s}"
                f" → column {column:3d}"
                f" → TF-IDF = {value:.4f}"
            )


# ============================================================
# STEP 7 — EXPLAIN FIRST SENTENCE
# ============================================================

print("\n" + "=" * 80)
print("STEP 7: FIRST SENTENCE")
print("=" * 80)

sentence_id = 0

print("\nSentence:")
print(sentences[sentence_id])

print("\nTF-IDF representation:")

for column, value in enumerate(emb_tfidf[sentence_id]):

    if value > 0:

        print(
            f"{feature_names[column]:15s}"
            f" → {value:.4f}"
        )


# ============================================================
# STEP 8 — SPECIFICALLY CHECK DELHI
# ============================================================

print("\n" + "=" * 80)
print("STEP 8: WHAT HAPPENS TO 'DELHI'?")
print("=" * 80)

delhi_column = tfidf.vocabulary_.get("delhi")

print("\nDelhi column index:", delhi_column)

print("\nDelhi TF-IDF value in every sentence:")

for i, sentence in enumerate(sentences):

    value = emb_tfidf[i][delhi_column]

    print(
        f"Sentence {i:2d}"
        f" → Delhi TF-IDF = {value:.4f}"
        f" → {sentence}"
    )


# ============================================================
# STEP 9 — UNDERSTAND TF
# ============================================================

print("\n" + "=" * 80)
print("STEP 9: TERM FREQUENCY (TF)")
print("=" * 80)

print("""
TF measures how frequently a word appears
inside a particular sentence.

Example:

"Delhi is the capital of India."

Delhi appears once.

TF(Delhi) = 1 occurrence
""")


# ============================================================
# STEP 10 — UNDERSTAND IDF
# ============================================================

print("\n" + "=" * 80)
print("STEP 10: INVERSE DOCUMENT FREQUENCY (IDF)")
print("=" * 80)

print("""
IDF measures how rare or common a word is
across all sentences.

Rare word:
    Higher IDF

Common word:
    Lower IDF

Therefore:

Rare + important word
        ↓
Higher TF-IDF

Common word
        ↓
Lower TF-IDF
""")


# ============================================================
# STEP 11 — SHOW ACTUAL IDF VALUES
# ============================================================

print("\n" + "=" * 80)
print("STEP 11: ACTUAL IDF VALUES")
print("=" * 80)

idf_values = tfidf.idf_

for i, word in enumerate(feature_names):

    print(
        f"{word:20s}"
        f" → IDF = {idf_values[i]:.4f}"
    )


# ============================================================
# STEP 12 — DELHI'S ACTUAL IDF
# ============================================================

print("\n" + "=" * 80)
print("STEP 12: DELHI'S IDF")
print("=" * 80)

print(
    f"\nDelhi → IDF = "
    f"{idf_values[delhi_column]:.4f}"
)

print(
    "\nDelhi appears in only one sentence, "
    "so it is relatively informative."
)


# ============================================================
# STEP 13 — RAW TF × IDF BEFORE NORMALIZATION
# ============================================================

print("\n" + "=" * 80)
print("STEP 13: TF × IDF")
print("=" * 80)

print("""
TF-IDF conceptually starts with:

TF-IDF = TF × IDF

The resulting vector is then normalized
by scikit-learn.

Therefore, the final displayed TF-IDF
values may be different from the raw TF × IDF.
""")


# ============================================================
# STEP 14 — FINAL SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("FINAL PIPELINE")
print("=" * 80)

print("""
Raw Sentence
     ↓
Tokenization
     ↓
Vocabulary Creation
     ↓
Term Frequency (TF)
     ↓
Inverse Document Frequency (IDF)
     ↓
TF × IDF
     ↓
Normalization
     ↓
Final TF-IDF Vector
""")
```

```python
from gensim.models import Word2Vec
from gensim.utils import simple_preprocess
from sklearn.metrics.pairwise import euclidean_distances, cosine_similarity
import numpy as np


# ============================================================
# SENTENCES
# ============================================================

sentences = [

    "Delhi is the capital of India.",
    "Tokyo is a major city in Japan.",
    "Paris is known as the city of lights.",
    "New York is one of the busiest cities in the world.",
    "London is famous for its historical landmarks.",
    "Sydney is a coastal city known for the Opera House.",
    "LLMs are used for text processing.",
    "Transformers help machines understand language.",
    "NLP models learn linguistic patterns.",
    "Tokenization converts text into numerical units.",
    "Language models generate human-like responses.",
    "Text classification assigns labels to sentences.",
    "CNNs are popular for image classification.",
    "Vision Transformers perform well on image tasks.",
    "Object detection identifies items in pictures.",
    "Image segmentation divides an image into meaningful regions.",
    "Face recognition systems detect and identify faces.",
    "Optical character recognition extracts text from images."
]


# ============================================================
# STEP 1 — INTRODUCTION TO WORD2VEC
# ============================================================

print("=" * 80)
print("STEP 1: INTRODUCTION TO WORD2VEC")
print("=" * 80)

print("""
Word2Vec learns numerical representations of words
from their surrounding context.

Unlike TF-IDF, Word2Vec does not assign one
dimension to every vocabulary word.

Instead, each word is represented by a
dense, fixed-dimensional vector.

Example:

    "Delhi"
        ↓
    [0.12, -0.34, 0.08, ..., 0.21]

The vector dimensions are learned from context.
""")


# ============================================================
# STEP 2 — PREPROCESSING AND TOKENIZATION
# ============================================================

print("\n" + "=" * 80)
print("STEP 2: TOKENIZATION")
print("=" * 80)

tokenized = [
    simple_preprocess(sentence)
    for sentence in sentences
]

for i, (sentence, tokens) in enumerate(
    zip(sentences, tokenized)
):

    print(f"\nSentence {i}")
    print("Original :", sentence)
    print("Tokens   :", tokens)


# ============================================================
# STEP 3 — UNDERSTAND THE CONTEXT WINDOW
# ============================================================

print("\n" + "=" * 80)
print("STEP 3: CONTEXT WINDOW")
print("=" * 80)

print("""
Word2Vec learns word representations using
the words that occur around them.

For example:

    Delhi is the capital of India

A word is associated with nearby words
within the specified context window.

window = 5

means Word2Vec considers up to five
words around the target word where available.

The central idea is:

    Similar contexts
          ↓
    Similar word representations
""")


# ============================================================
# STEP 4 — TRAIN WORD2VEC MODEL
# ============================================================

print("\n" + "=" * 80)
print("STEP 4: TRAIN WORD2VEC MODEL")
print("=" * 80)

w2v_model = Word2Vec(
    sentences=tokenized,
    vector_size=50,     # embedding dimension
    window=5,           # context window
    min_count=1,        # keep words appearing at least once
    workers=4,
    sg=1,               # Skip-gram
    epochs=200
)

print("""
Word2Vec model trained successfully.

Parameters:

Embedding dimension = 50
Context window      = 5
Minimum word count  = 1
Training algorithm  = Skip-gram
Epochs              = 200
""")


# ============================================================
# STEP 5 — WORD VOCABULARY
# ============================================================

print("\n" + "=" * 80)
print("STEP 5: WORD2VEC VOCABULARY")
print("=" * 80)

print("\nWords learned by Word2Vec:\n")

for i, word in enumerate(w2v_model.wv.index_to_key):

    print(
        f"{i:3d} → {word}"
    )

print(
    "\nVocabulary size:",
    len(w2v_model.wv)
)


# ============================================================
# STEP 6 — WORD EMBEDDING DIMENSION
# ============================================================

print("\n" + "=" * 80)
print("STEP 6: WORD EMBEDDING DIMENSION")
print("=" * 80)

print(
    "\nEmbedding dimension:",
    w2v_model.vector_size
)

print("""
Every word is represented using a
50-dimensional dense vector.

Therefore:

    Word
      ↓
    50 numerical values
""")


# ============================================================
# STEP 7 — SHOW A WORD VECTOR
# ============================================================

print("\n" + "=" * 80)
print("STEP 7: WORD VECTOR")
print("=" * 80)

word = "delhi"

if word in w2v_model.wv:

    vector = w2v_model.wv[word]

    print(f"\nWord: {word}")

    print("\nVector:")

    print(
        np.round(vector, 4)
    )

    print(
        "\nVector shape:",
        vector.shape
    )


# ============================================================
# STEP 8 — SHOW SELECTED WORD VECTORS
# ============================================================

print("\n" + "=" * 80)
print("STEP 8: SELECTED WORD EMBEDDINGS")
print("=" * 80)

selected_words = [
    "delhi",
    "india",
    "tokyo",
    "japan",
    "city",
    "language",
    "transformers",
    "image"
]

for word in selected_words:

    if word in w2v_model.wv:

        print(f"\n{word}")

        print(
            np.round(
                w2v_model.wv[word],
                4
            )
        )


# ============================================================
# STEP 9 — WORD SIMILARITY
# ============================================================

print("\n" + "=" * 80)
print("STEP 9: WORD SIMILARITY")
print("=" * 80)

word = "city"

if word in w2v_model.wv:

    print(
        f"\nWords most similar to '{word}':\n"
    )

    similar_words = w2v_model.wv.most_similar(
        word,
        topn=5
    )

    for similar_word, score in similar_words:

        print(
            f"{similar_word:20s}"
            f" → similarity = {score:.4f}"
        )


# ============================================================
# STEP 10 — WORD VECTOR GEOMETRY
# ============================================================

print("\n" + "=" * 80)
print("STEP 10: WORD VECTOR GEOMETRY")
print("=" * 80)

print("""
Word2Vec places words in a vector space.

Conceptually:

    Similar context
          ↓
    Similar vector direction
          ↓
    Higher cosine similarity

Therefore:

    Word meaning
          ↓
    Distributional information
          ↓
    Numerical representation
          ↓
    Geometric relationships
""")


# ============================================================
# STEP 11 — BUILD SENTENCE REPRESENTATION
# ============================================================

print("\n" + "=" * 80)
print("STEP 11: BUILD SENTENCE REPRESENTATIONS")
print("=" * 80)

print("""
Word2Vec produces word embeddings.

It does not directly produce a
sentence embedding.

To obtain a simple sentence representation,
we average the word vectors.

For a sentence containing n words:

        1
    s = ─── Σ v(w_i)
        n
""")


# ============================================================
# STEP 12 — SENTENCE VECTOR FUNCTION
# ============================================================

print("\n" + "=" * 80)
print("STEP 12: SENTENCE VECTOR FUNCTION")
print("=" * 80)


def sentence_vector(tokens, model):

    vec = np.zeros(
        model.vector_size
    )

    count = 0

    for word in tokens:

        if word in model.wv:

            vec += model.wv[word]

            count += 1

    if count > 0:

        return vec / count

    else:

        return vec


print("""
Sentence vector function created.

Each sentence will be converted into
one 50-dimensional vector.
""")


# ============================================================
# STEP 13 — CREATE SENTENCE REPRESENTATIONS
# ============================================================

print("\n" + "=" * 80)
print("STEP 13: SENTENCE REPRESENTATIONS")
print("=" * 80)

emb_w2v = np.array(
    [
        sentence_vector(
            tokens,
            w2v_model
        )
        for tokens in tokenized
    ]
)

print(
    "\nWord2Vec Sentence Representation Matrix:\n"
)

print(
    np.round(
        emb_w2v,
        4
    )
)

print(
    "\nMatrix shape:",
    emb_w2v.shape
)

print("""
Rows    = sentences
Columns = embedding dimensions

Therefore:

    18 sentences × 50 dimensions
""")


# ============================================================
# STEP 14 — SHOW FIRST SENTENCE REPRESENTATION
# ============================================================

print("\n" + "=" * 80)
print("STEP 14: FIRST SENTENCE REPRESENTATION")
print("=" * 80)

sentence_id = 0

print("\nSentence:")
print(sentences[sentence_id])

print("\nTokens:")
print(tokenized[sentence_id])

print("\nSentence vector:")

print(
    np.round(
        emb_w2v[sentence_id],
        4
    )
)

print(
    "\nVector shape:",
    emb_w2v[sentence_id].shape
)


# ============================================================
# STEP 15 — SHOW HOW SENTENCE VECTOR IS CREATED
# ============================================================

print("\n" + "=" * 80)
print("STEP 15: WORD VECTORS → SENTENCE VECTOR")
print("=" * 80)

tokens = tokenized[0]

print("\nSentence:")
print(sentences[0])

print("\nWords contributing to the sentence vector:\n")

for word in tokens:

    if word in w2v_model.wv:

        print(
            f"{word:15s} → "
            f"{np.round(w2v_model.wv[word], 4)}"
        )

print("""
The sentence representation is obtained
by averaging these word vectors.
""")


# ============================================================
# STEP 16 — EUCLIDEAN DISTANCE
# ============================================================

print("\n" + "=" * 80)
print("STEP 16: EUCLIDEAN DISTANCE")
print("=" * 80)

D_w2v = euclidean_distances(
    emb_w2v
)

print(
    "\nEuclidean Distance Matrix:\n"
)

print(
    np.round(
        D_w2v,
        4
    )
)

print("""
Euclidean distance between two sentence
vectors is:

d(x,y) =
sqrt(Σ(x_i - y_i)^2)

Smaller distance
        ↓
Geometrically closer vectors
""")


# ============================================================
# STEP 17 — SUM OF SQUARED DISTANCES
# ============================================================

print("\n" + "=" * 80)
print("STEP 17: SUM OF SQUARED DISTANCES (SSD)")
print("=" * 80)

SSD_w2v = np.sum(
    D_w2v ** 2
)

print(
    "\nSum of Squared Distances (SSD):",
    SSD_w2v
)

print("""
SSD is the sum of squared pairwise
Euclidean distances in the distance matrix.

It provides a single numerical quantity
representing the overall pairwise dispersion
of the sentence representations.
""")


# ============================================================
# STEP 18 — COSINE SIMILARITY
# ============================================================

print("\n" + "=" * 80)
print("STEP 18: COSINE SIMILARITY")
print("=" * 80)

cos_w2v = cosine_similarity(
    emb_w2v
)

print(
    "\nCosine Similarity Matrix:\n"
)

print(
    np.round(
        cos_w2v,
        4
    )
)

print("""
Cosine similarity measures the angle
between two vectors.

              x · y
cos(θ) = ─────────────
          ||x|| ||y||

Higher value
      ↓
More similar vector directions

Lower value
      ↓
Less similar vector directions
""")


# ============================================================
# STEP 19 — COMPARE TWO SENTENCES
# ============================================================

print("\n" + "=" * 80)
print("STEP 19: COMPARE TWO SENTENCES")
print("=" * 80)

sentence_a = 0
sentence_b = 1

print("\nSentence A:")
print(sentences[sentence_a])

print("\nSentence B:")
print(sentences[sentence_b])

print(
    "\nEuclidean distance:",
    round(
        D_w2v[sentence_a][sentence_b],
        4
    )
)

print(
    "\nCosine similarity:",
    round(
        cos_w2v[sentence_a][sentence_b],
        4
    )
)


# ============================================================
# STEP 20 — FIND MOST SIMILAR SENTENCE
# ============================================================

print("\n" + "=" * 80)
print("STEP 20: MOST SIMILAR SENTENCE")
print("=" * 80)

sentence_id = 0

similarities = cos_w2v[sentence_id].copy()

# Ignore similarity with itself
similarities[sentence_id] = -1

most_similar_id = np.argmax(
    similarities
)

print("\nQuery sentence:")
print(sentences[sentence_id])

print("\nMost similar sentence:")
print(sentences[most_similar_id])

print(
    "\nCosine similarity:",
    round(
        similarities[most_similar_id],
        4
    )
)


# ============================================================
# STEP 21 — FINAL WORD2VEC PIPELINE
# ============================================================

print("\n" + "=" * 80)
print("STEP 21: FINAL WORD2VEC PIPELINE")
print("=" * 80)

print("""
Raw Sentences
      ↓
Tokenization
      ↓
Context Windows
      ↓
Word2Vec Training
      ↓
Word Embeddings
      ↓
50-dimensional Word Vectors
      ↓
Mean Pooling
      ↓
Sentence Representations
      ↓
Distance / Similarity
      ↓
Euclidean Distance
      +
Cosine Similarity
""")
```
