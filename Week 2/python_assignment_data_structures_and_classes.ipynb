### Question 20: Creating a Class and Objects
1. Define a class named `Car`.
2. The class should have an `__init__` constructor method that initializes three instance attributes: `make`, `model`, and `year`.
3. Define a method named `display_info` inside the class that returns a formatted string like: `"2020 Toyota Corolla"`.
4. Create an object (instance) of `Car` with make `"Toyota"`, model `"Corolla"`, and year `2020`.
5. Print the result of calling the `display_info` method on the object.

````python
# Write your code here
class car:
    def __init__(self, make:str, model:str, year:int):
        self.make = make
        self.model = model
        self.year = year

    def display_info(self):
        print(f"{self.year} {self.make} {self.model}")

mycar = car("Toyota", "Camry", 2012)
mycar.display_info()
        
````

#### **Expected Output:**
```text
2020 Toyota Corolla
```

### Question 21: Inheritance and Overriding
1. Define a class `ElectricCar` that inherits from `Car`.
2. It should have an `__init__` constructor that accepts `make`, `model`, `year`, and an additional attribute `battery_capacity` (in kWh). Use `super()` to initialize the base class attributes.
3. Override the `display_info` method to append the battery capacity, returning a string like: `"2023 Tesla Model 3 (75 kWh battery)"`.
4. Create an instance of `ElectricCar` with make `"Tesla"`, model `"Model 3"`, year `2023`, and battery capacity `75`.
5. Print the result of calling `display_info` on the instance.

````python
# Write your code here
class ElectricCar(car):
    # 2. Constructor accepting make, model, year, and battery_capacity
    def __init__(self, make: str, model: str, year: int, battery_capacity: int):
        # Use super() to initialize base class attributes
        super().__init__(make, model, year)
        self.battery_capacity = battery_capacity

    # 3. Override display_info to return the specific string format
    def display_info(self) -> str:
        return f"{self.year} {self.make} {self.model} ({self.battery_capacity} kWh battery)"

my_EV = ElectricCar("Tesla", "model3", 2023, "75 kwh")
my_EV.display_info()
````

#### **Expected Output:**
```text
2023 Tesla Model 3 (75 kWh battery)
```
