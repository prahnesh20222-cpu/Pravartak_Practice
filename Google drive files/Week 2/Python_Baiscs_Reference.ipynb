# **Custom Functions**

```python
def calculate_interest(principal: float, duration: int, rate: float = 0.05) -> float:
  # I = PNR
  interest = principal * rate * duration
  return interest



principal = float(input("Enter the Amount :"))
rate = 0.09 # 9%
duration_1 = 5
duration_2 = 10
duration_3 = 15

interest_1 = calculate_interest(principal, duration_1, rate)

interest_2 = calculate_interest(principal, duration_2, rate)

interest_3 = calculate_interest(principal, duration_2)

print(f"your interest with duration {duration_1} : {interest_1}")
print(f"your interest with duration {duration_2} : {interest_2}")
print(f"your interest with duration {duration_3} : {interest_3}")
```

```python
class Student:
  name:str
  age:int
  email:str
  no_subjects:int
  marks:list
  total_marks:int
  average_marks:float

  def input_values(self, inputted_name:str, inputted_age:int, inputted_email:str, inputted_marks:list, inputted_no_subjects:int = 5):
    self.name = inputted_name
    self.age = inputted_age
    self.email = inputted_email
    self.no_subjects = inputted_no_subjects
    self.marks = inputted_marks

    self.calculate_total_marks()
    self.calculate_average_marks()


  def calculate_total_marks(self):
    self.total_marks = sum(self.marks)

  def calculate_average_marks(self):
    self.average_marks =  self.total_marks / self.no_subjects


  def print_my_object(self):
    print("\n\n")
    print("=================================")
    print("The Student information....!")
    print(f"Name : {self.name}")
    print(f"Age : {self.age}")
    print(f"Email : {self.email}")
    print(f"No of Subject : {self.no_subjects}")
    print(f"Marks : {self.marks}")
    print(f"Total Marks Earned : {self.total_marks}")
    print(f"Avg Marks : {self.average_marks}")

```

```python
class Student:
  name:str
  age:int
  email:str
  no_subjects:int
  marks:list
  total_marks:int
  average_marks:float

  def __init__(self, inputted_name:str, inputted_age:int, inputted_email:str, inputted_marks:list, inputted_no_subjects:int = 5):
    self.name = inputted_name
    self.age = inputted_age
    self.email = inputted_email
    self.no_subjects = inputted_no_subjects
    self.marks = inputted_marks

    self.calculate_total_marks()
    self.calculate_average_marks()


  def calculate_total_marks(self):
    self.total_marks = sum(self.marks)

  def calculate_average_marks(self):
    self.average_marks =  self.total_marks / self.no_subjects


  def print_my_object(self):
    print("\n\n")
    print("=================================")
    print("The Student information....!")
    print(f"Name : {self.name}")
    print(f"Age : {self.age}")
    print(f"Email : {self.email}")
    print(f"No of Subject : {self.no_subjects}")
    print(f"Marks : {self.marks}")
    print(f"Total Marks Earned : {self.total_marks}")
    print(f"Avg Marks : {self.average_marks}")

```

```python
name = input("Enter the student name :")
age = int(input("Enter the student age :"))
email = input("Enter the student email :")
no_subjects = int(input("Enter the no subjects :"))
marks = []
for i in range(no_subjects):
  mark = int(input(f"Enter the marks of subject {i + 1} :"))
  marks.append(mark)

stu1 = Student(inputted_age=age, inputted_marks=marks, inputted_name=name, inputted_email=email, inputted_no_subjects=no_subjects)
stu1.print_my_object()
```
