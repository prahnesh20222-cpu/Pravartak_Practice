```python
import asyncio
import time

async def test_speed_and_logic():
    prompt = """Act as a Principal Systems Architect. I have a monolithic service that needs to be refactored into an asynchronous event-driven microservices architecture using Python, FastAPI, and Redis.

Here are the strict, conflicting constraints you must simultaneously satisfy in your design:
1. The API gateway must implement a sliding-window rate limiter directly inside Redis using a Lua script to ensure atomicity, but the fallback mechanism during a Redis cluster outage must fail-open for authenticated enterprise tiers and fail-closed for anonymous tiers without introducing middleware latency higher than 2ms.
2. Design a custom LangGraph state-reduction node that compresses conversation history dynamically when token length crosses 75% of the context window, ensuring that system prompts and the last 3 user-assistant turns are mathematically immune to truncation.
3. Provide the complete, production-ready Python code for both components. Do not use pseudo-code, do not truncate implementation details with comments like "# add logic here", and ensure all type hints and Pydantic v2 models are fully realized.'"""

    models_to_test = ["gpt-4o", "gpt-4o-mini"]

    for model in models_to_test:
        print(f"==================================================")
        print(f"Testing Model: {model}")
        print(f"==================================================\n")

        # Start the timer
        start_time = time.time()

        response = await client.chat.completions.create(
            model=model,
            messages=[{"role": "user", "content": prompt}],
            stream=False,
        )

        # End the timer
        end_time = time.time()
        execution_time = end_time - start_time

        output = response.choices[0].message.content
        print(output)

        # Print the execution time
        print(f"\n[⏱️ Time taken: {execution_time:.2f} seconds]\n\n")

# Run the test
await test_speed_and_logic()
```
