create task function to create a background task

```python
async def send_data_db():
    print("Sending data to the database...")
    await asyncio.sleep(10)  # Simulate a delay for sending data
    print("Data sent to the database.")
    return 100

async def my_workflow():
    print("Starting workflow...")
    task = asyncio.create_task(send_data_db())
    print("Workflow continue.")
    await asyncio.sleep(5)  # Simulate some other work being done       
    print("Workflow finished.")
    return 200

await my_workflow()

#logging is a good use case for background task
#redis cache clearance is a good use case for background task

#Can you call a regular function from an async function? If yes, how?
#Yes, you can call a regular (synchronous) function from an async function. You simply call the synchronous function as you would in a normal context. However, keep in mind that calling a synchronous function will block the event loop if it takes a long time to execute, which can affect the performance of your asynchronous code.

#Does task function return a value? If yes, how can you get the value?
```
