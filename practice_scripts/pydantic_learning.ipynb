```python
class User(BaseModel):
    name:str = Field(min_length = 3, max_length=50)
    age:int = Field(ge=0, le=120, default=18)
    occupation:str = Field(min_length = 3, max_length=50, description="The user's occupation")
    hobbies:list[str]
```

```python
from pydantic import BaseModel, field_validator,Field

class User(BaseModel):
    age: int

    @field_validator("age") # this is a decorator to identify the checl_age as a method to use for validation at the time of object creation.
    @classmethod
    def check_age(cls, value):
        if value < 0:
            raise ValueError("Age cannot be negative")
        return value
```

```python
#Combine Field and fioeld_validator
class User(BaseModel):
    age : int = Field(ge=0, le=120, default=18)

    @field_validator("age") # this is a decorator to identify the checl_age as a method to use for validation at the time of object creation.
    @classmethod
    def check_age(cls, value):
        if value %2 != 0:
            raise ValueError("Age cannot be odd number")
        return value
```

```python
john_age = User(age = 20)
john_age = User(age = '20')
#john_age = User(age = 21)
john_age = User()
```
