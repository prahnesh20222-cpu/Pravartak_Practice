```python
def hello():
    print("hello world")
hello()
```

Converting the function to async requires the key word as below. 

```python
async def hello():
    print("hello world")
# Calling the function in the usual manner will not print output.it will return a coroutine object instead.
# <coroutine object hello at 0x00000247ADBABB80>
hello()

await hello()  # This will print "hello world"
```

```python
import asyncio

async def hello():
    print("hello world")

async def howdy():
    print("howdy")

async def fine():
    print("I'm fine!")

await asyncio.gather(hello(), howdy(), fine()) 
 # This will print "hello world", "howdy", and "I'm fine!" in any order.
```

**ASYNC SLEEP**

```python
async def hello():
    await asyncio.sleep(5)
    print("hello world")

async def howdy():
    await asyncio.sleep(2)
    print("howdy")

async def fine():
    await asyncio.sleep(1)
    print("I'm fine!")

await asyncio.gather(hello(), howdy(), fine()) 
 # This will print "hello world", "howdy", and "I'm fine!" in reverse order.
```

** CREATE TASK**

```python
async def send_data_db():
    print("Sending data to db is initiated..")
    await asyncio.sleep(10)
    print("Data sent to db succesfully..")

async def my_workflow():
    print("Processing the data..")
    task = asyncio.create_task(send_data_db())
    print("continue with rest of the workflow..")
    await asyncio.sleep(3)
    print("Data processing is completed..")

await my_workflow()


```
